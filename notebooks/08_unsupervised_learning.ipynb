# 08_unsupervised_learning.ipynb

**Objective:** Cluster and reduce dimensionality without using the supervised target.

All observations are synthetic; avoid real-world claims.

```python
from pathlib import Path
ROOT = Path.cwd()
if not (ROOT / 'data').exists(): ROOT = ROOT.parent
assert (ROOT / 'data').exists(), 'Run Jupyter from the repository root or notebooks folder'
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from sklearn.model_selection import train_test_split
health = pd.read_csv(ROOT / 'data/clean/healthcare_clean.csv')
retail = pd.read_csv(ROOT / 'data/clean/retail_clean.csv')
print('Healthcare', health.shape, 'Retail', retail.shape)
```

## Reproducible demonstration

Interpret results in the context of the synthetic generating process.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
cols=['age','bmi','biomarker_level','imaging_score','inflammation_marker'];Z=StandardScaler().fit_transform(health[cols]);labels=KMeans(n_clusters=3,n_init=10,random_state=42).fit_predict(Z)
coords=PCA(n_components=2).fit_transform(Z);fig,ax=plt.subplots();ax.scatter(coords[:,0],coords[:,1],c=labels,s=10);ax.set_title('Unsupervised synthetic patient clusters');plt.show()
print('Cluster sizes',np.bincount(labels))
```
