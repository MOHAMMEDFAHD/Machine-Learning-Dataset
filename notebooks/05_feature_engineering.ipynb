## Reproducible demonstration

Interpret results in the context of the synthetic generating process.

```python
retail['date']=pd.to_datetime(retail.date);retail=retail.sort_values(['product_id','date']);retail['weekday']=retail.date.dt.dayofweek
retail['lag1']=retail.groupby('product_id').units_sold.shift(1)
retail['rolling7']=retail.groupby('product_id').units_sold.transform(lambda x:x.shift(1).rolling(7,min_periods=1).mean())
print(retail[['product_id','date','units_sold','lag1','rolling7']].head(12))
```
