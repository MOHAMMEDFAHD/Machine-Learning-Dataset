```python
from pathlib import Path
ROOT = Path.cwd()
if not (ROOT / 'data').exists(): ROOT = ROOT.parent
assert (ROOT / 'data').exists(), 'Run Jupyter from the repository root or notebooks folder'
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from sklearn.model_selection import train_test_split
health = pd.read_csv(ROOT / 'data/clean/healthcare_clean.csv')
retail = pd.read_csv(ROOT / 'data/clean/retail_clean.csv')
print('Healthcare', health.shape, 'Retail', retail.shape)
```

## Reproducible demonstration

Interpret results in the context of the synthetic generating process.

```python
ages=health.age.to_numpy();print('Mean age',ages.mean(),'Over 60',np.sum(ages>60))
print('First five',ages[:5]);print('Broadcasting example',np.round((ages[:5]-ages.mean())/ages.std(),2))
rng=np.random.default_rng(42);print('Reproducible random sample',rng.choice(ages,5,replace=False))
```
