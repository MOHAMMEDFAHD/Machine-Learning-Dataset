```python
from pathlib import Path
ROOT = Path.cwd()
if not (ROOT / 'data').exists(): ROOT = ROOT.parent
assert (ROOT / 'data').exists(), 'Run Jupyter from the repository root or notebooks folder'
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from sklearn.model_selection import train_test_split
health = pd.read_csv(ROOT / 'data/clean/healthcare_clean.csv')
retail = pd.read_csv(ROOT / 'data/clean/retail_clean.csv')
print('Healthcare', health.shape, 'Retail', retail.shape)
```

## Reproducible demonstration

Interpret results in the context of the synthetic generating process.

```python
import sys
sys.path.insert(0,str(ROOT/'src'))
from model_examples import healthcare
results,splits=healthcare();print('Stratified splits',splits);print('Validation metrics',results)
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay
from sklearn.linear_model import LogisticRegression
from model_examples import pipeline
X=health.drop(columns=['patient_id','cancer_status']);y=health.cancer_status
tr,tmp=train_test_split(np.arange(len(y)),test_size=.30,stratify=y,random_state=42)
va,te=train_test_split(tmp,test_size=.5,stratify=y.iloc[tmp],random_state=42)
model=pipeline(X.iloc[tr],LogisticRegression(max_iter=2000)).fit(X.iloc[tr],y.iloc[tr]);prob=model.predict_proba(X.iloc[va])[:,1]
ConfusionMatrixDisplay.from_predictions(y.iloc[va],prob>=.5);plt.show()
print('Sensitivity',((prob>=.5)&(y.iloc[va].to_numpy()==1)).sum()/sum(y.iloc[va]==1))
```
